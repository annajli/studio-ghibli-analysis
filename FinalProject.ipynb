# Interpretation (4)

Describe something interesting about your corpus that you discovered during the process of completing this assignment.

At a minumum, use 250 words, but you may use more. You may also add images if you'd like.

I hypothesized that I would see more clear directorial differences between Miyazaki's and Takahata's films, but that didn't really turn out to be the case, throughout the various lines of text analysis. It's interesting to note that there isn't much overlap between similar topics for the Takahata films, such that our topic x film heatmap in Riff 3 corroborates what we saw in the dendrogram in Riff 1, that the clustering for Takahata's films is not very strong. The limitations here though are that we're looking at a small sample size for Takahata's films (and the corpus is small to begin with), and the nature of films, such that directorial influence shows up not solely in screenplay, but in cinematography / animation sequencing / narrative storytelling, that aren't captured solely in text.

What's really interesting though, is how consistent Miyazaki's thematic vocabulary turns out to be. Topics around flight, adventure, and spiritual transformation show up repeatedly across films that span almost thirty years, from Nausicaa (1984) to The Wind Rises (2013), pointing to a recognizable authorial voice that holds even as his storytelling evolves. 

I also found it really cool that from the sentiment analysis, we found that trust is consistently the strongest positive emotion across speakers and films. Given that Studio Ghibli is often associated with wonder and bittersweet melancholy, it demonstrates that the dialogue very much focuses on relationships and characters depending on each other. 

Future work could look more into differences between source material, because we sort of pulled the screenplays from wherever they were readily available, regardless of whether they were fan-translations of the original Japanese subtitles / official English subtitles / based off of English dub. I'm also curious as to what text analysis would look like if we were using the original Japanese screenplays, and the differences in techniques that we would need to use, given an entirely different world view.
